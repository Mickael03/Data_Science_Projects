## Dados ausentes

```python
# Identificação da posição dos valores ausentes
ausentes = [] # Armazenar a posição da cada valor ausente
for index in range(len(df.index)):
    for columns in range(len(df.columns)):
        if pd.isna(df.iloc[index,columns]):
            ausentes.append((index,columns))

print(f'Quantidade de valores ausentes: {len(ausentes)}')

"""
index = list(df.dropna().index) # Identifica os indexs dos valores ausentes
"""
```

### Explicar a diferença entre `None` e `NaN`

### Diferença entre `==` e `is`

```python
# Excluindo os valores ausentes, exclusão de toda as amostras com valores ausentes
for i in ausentes:
    try:  # Bloco try-except é 
        df.drop(i[0], inplace=True)
    except:
        continue
```

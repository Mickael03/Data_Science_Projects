O fluxograma acima representa o modelo **CRISP-DM** (acrônimo para *Cross-Industry Standard Process for Data Mining*) metodologia utilizada para o desenvolvimento de projetos de Data Science e Machine Learning.

Ele divide qualquer projeto de dados em **6 fases principais**, que não são estritamente lineares, mas sim um ciclo contínuo e iterativo.

1. **Compreensão do Negócio (Business Understanding)**: Antes de tocar em qualquer dado ou escrever uma linha de código, você precisa entender o problema.
2. **Compreensão dos Dados (Data Understanding)**: Coleta inicial dos dados, exploração (Análise Exploratória de Dados) e verificação da qualidade.
3. **Preparação dos Dados (Data Preparation)**: Os dados brutos são transformados no formato para serem usado pelo modelo.
4. **Modelagem (Modeling)**: Seleção das técnicas apropriadas de acordo com o problema (classificação, regressão, clusterização).
5. **Avaliação (Evaluation)**: Avaliação dos resultados do modelo
6. **Implantação (Deployment)**: O modelo é colocado em produção

A maior força do CRISP-DM é que ele é um *processo iterativo*.

## Variável Dummy

Uma variável dummy é utilizada para representar variáveis categóricas por meio de valores binários (0 e 1). Nesse caso, empregaremos esse tipo de variável para codificar as indicações de sexo da seguinte maneira:
|Sexo|Valor|
|----|-----|
|female|0|
|male|1|

```python
# Trocar de valores para variável categórica 'Sex'
for i in range(len(df)):
    if df.loc[i,'Sex'] == 'female':
        df.loc[i,'Sex'] = 0
    else:
        df.loc[i,'Sex'] = 1

# Mostra o dataframe
df.head()
```

Para `Embarked` transformaremos em uma variável dummy, como `Embarked` apresenta três categorias devemos criar duas variáveis dummys:
1. `Cherbourg` com valores 0 ou 1, sendo 1 quando `Embarked` tivesse sindo classificado como `C` e 0 quando não recebeu esse valor.
2. `Queenstown` com valor 0 ou 1, sendo 1 quando `Embarked` tivesse sindo classificado como `Q`e 0 quando não recebeu esse valor.
3. Quando `Cherbourg` e `Queenstown` receberem valor 0, saberemos que `Embarked` foi classificado como `S` (Southampton).

Quando temos $k$ categorias, são necessários $k-1$ categorias variáveis dummy. Isso é necessários para evitamos ***Multicolinearidade*** que é quando duas ou mais variáveis independentes apresentam uma alta correlação.

****Tabela de Representação****
|Categorias|Cherbourg|Queenstown|
|:--------:|:-------:|:--------:|
|`C`|1|0|
|`Q`|0|1|
|`S`|0|0|

```python
# Lista de lista para transformar 'Embarked' em um variável Dummy
embarked = []

# Determinar o valor de cada variável Dummy
for index in range(len(df)):
    if df.loc[index,'Embarked'] == 'C':
        embarked.append([1,0])
    elif df.loc[index,'Embarked'] == 'Q':
        embarked.append([0,1])
    else:
        embarked.append([0,0])

# Dataframe como o atributo 'Embarked' transformando em um variável dummy de acordo com suas categórias
embarked = pd.DataFrame(embarked, columns=['Cherbourg','Queenstown'])

# Mostra o final dataframe
embarked.tail()
```

```python
# Concatenando os dois dataframe embarked e df
df = pd.concat([df,embarked], axis=1)

# Mostrar o Dataframe
df
```

```python
# Excluir o Atributo 'Embarked' que não é mais necessário foi trocado pelos atributos 'Cherbourg', 'Queenstown' e 'Southampton'
df.drop(columns=['Embarked'], inplace=True)

# Mostra o dataframe
df
```

No momentos não iremos trabalha como dados ausentes, deixaremos isse detalhe para um próximo notebook.

```python
def ausentes(DataFrame, quantidade=False):
    # Identificação da posição dos valores ausentes
    '''
    Paramêtros:
        DataFrame: DataFrame do Pandas
        quantidade (bool): True mostra indica a quantidade de dados ausentes

    Return:
        list: Uma lista de tuplas onde temos (linha,coluna) de cada valor ausente
        int: Quantidade de dados ausentes
    '''
    ausentes = [] # Armazenar a posição da cada valor ausente
    for index in range(len(df.index)):
        for columns in range(len(df.columns)):
            if pd.isna(df.iloc[index,columns]):
                ausentes.append((index,columns))

    return len(ausentes) if quantidade else ausentes
```

```python
# Lista de valores ausentes
valores_ausentes = ausentes(df)

# Indica a quantidade de valores ausentes
print(f'Quantidade de valores ausentes: {len(valores_ausentes)}')

# Exclusão de toda as amostras com valores ausentes restantes
for posicao in valores_ausentes:
    try:  # Bloco try-except é para evitar quebrar de código já que tera index repetidos
        df.drop(posicao[0], inplace=True)
    except:
        continue

# Atualização dos index
df.index = df.index

# Mostrar o dataframe
df.head()
```

# Transformação dos dados

A transformação dos dados é essencial para garanti a criação de modelos de predição coerentes já que determinados tipos de modelos aceitam somente determinando padrão estatístico de dados inseridos.

## Normalização

Sendo $X$ o valor que será transformado, $X_{Min}$ o menor valor da série dos dados e $X_{Max}$ o maior valor da série de dados, a normalização do valor $X$ será dado por:
    
<center> $X^{'}=\displaystyle\frac{X-X_{Min}}{X_{Max}-X_{Min}}$ <center>

```python
# Dataframe com valores normalizados
normalizados = pd.DataFrame()
```

```python
# Inserir valores normalizados
for i in df.columns:
    i_max = df[i].max() # Maior valor da i-coluna
    i_min = df[i].min() # Menor valor da i-coluna
    normalizados[i] = [(valor - i_min)/(i_max - i_min) for valor in df[i]] # Arry com dados da i-coluna
```

```python
# Salvado os dados normalizados e padronizados
normalizados.to_csv("Dados\\normalizados.csv", index = False)
normalizados.head(2)
```

## Padronização

Sendo $X$ o valor que será transfornado, $\mu$ a média e $\sigma$ o desvio padrão dos dados, a padronização do valor $X$ será dados por:

<center>$z = \displaystyle\frac{X - \mu}{\sigma}$<center>

Onde o valor $z$ é chamado de z-score.

```python
# Dataframe com valores padronizados
padronizados = pd.DataFrame()
```

```python
# Inserir valores padronizados
for i in df.columns:
    media = df[i].mean() # Média dos valores na i-coluna
    desvio_padrao = df[i].std() # Desvio padrão dos valores na i-coluna
    padronizados[i] = [(valor - media)/desvio_padrao for valor in df[i]] # Arry com dados da i-coluna
```

```python
# Salvado os dados padronizados
padronizados.to_csv("Dados\\padronizados.csv", index = False)
padronizados.head(2)
```

# Treinamento

O Teorema No Free Lunch afirma que, na média de todos os problemas possíveis, todos os algoritmos de aprendizagem apresentam basicamente o mesmo desempenho. Em outras palavras, qualquer algoritmo pode se mostrar excelente em um determinado problema, porém medíocre em outro. Dessa forma, na média, todos terão o mesmo desempenho quando consideramos todos os problemas possíveis aos quais podem ser aplicados.

**Os modelos de machine learning**

1. **Supervisionado**: Cada amostra de entrada dos dados rotulados de entrada estão associado a uma saída
2. **Não Supervisionado**: As amostras não tem rótulos permitindo que o modelo desenvolva seu próprio padrão de associação
3. **Aprendizagem por Reforço**: O algoritmo de machine learning aprender por tentativa e erro

No momentos estamos trabalho como dados bem estruturados e como temos valores especifícos de saída trabalharemos com modelos de machine leraning supervisionados

```python
# Importação dos modelos para treino
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
```

```python
# Salvar os dados dos modelos treinados
from joblib import dump

# Tratamenoto de datas
from datetime import date
```

Impacto da transformação de dados em cada modelo (Padronizados, Normalizados ou Dados Brutos):

***Modelos***

1. Logistic Regression:
> Esse tipo de modelo não apresenta grandes mudanças de desempenho entre os dados normalizados e padronizados, entretanto, os dados normalizados tornam mais fácil fazermos uma comparação entre os coeficientes do modelo.

2. SVM:
> Exige a padronização ou normalização dos dados, já que depende de cálculos de distância dos atributos. Sem a aplicação de uma das transformações, o modelo corre o risco de ficar enviesado como consequência das variáveis com maior escala.

3. K-Nearest Neighbors:
> Sensível à escala das variáveis por ser um modelo que depende basicamente da distância entre os atributos. Assim como o SVM, a normalização é mais aconselhável por restringir a dominância dos valores de cada variável.

4. Decision Tree e Random Forest:
> São indiferentes a qualquer uma das transformações nos dados por utilizarem thresholds (valores de corte), analisando cada uma das variáveis e dividindo-as em pequenos nós para processar e tomar a decisão.

```python
# Criar instâncias dos modelos
logistic_regression = LogisticRegression()
decision_tree = DecisionTreeClassifier(random_state=31)
random_forest = RandomForestClassifier(random_state=31)
knn = KNeighborsClassifier()
svm = SVC()
```

```python
# Dicionario com os modelos
modelos_dict = { "logistic_regression": logistic_regression,
                 "decision_tree": decision_tree,
                 "random_forest": random_forest,
                 "knn": knn,
                 "svm": svm
}
```

```python
# Treinamentos dos modelos
modelos = {}

# Dados para treino
x_treino = padronizados[padronizados.columns]
y_treino = survived["Survived"]

# Treinamento e Inserção da informação
for nome, modelo in modelos_dict.items():
    modelo.fit(x_treino, y_treino)
    modelos[nome] = { "instancia": modelo,
                      "data_de_Treinamento": date.today().isoformat()
}
```

```python
# Salvando todos os modelos e metadados em um único arquivo
dump(modelos, "modelos_Parte_01.joblib")
```

# Avaliação

```python
# Métricas de avaliação
from sklearn.metrics import precision_score # Valor Preditivo Positivo
from sklearn.metrics import recall_score # Sensibilidade
from sklearn.metrics import f1_score # F1-Score
```

```python
# Leitura do arquivo de teste
teste = pd.read_csv("dados/test.csv")
survived = pd.read_csv("dados/gender_submission.csv")
```

```python
# Limpeza da base de dados para avaliaçõa
teste = teste.drop(columns=["Name", "Ticket", "Cabin"])               # Excluir os atributos que não foram utilizandos para modelagem
teste = teste.dropna()                                                # Excluir amostras que tenham algum valor ausente
teste["Sex"] = teste["Sex"].replace({"male": 1, "female": 0})         # Transformação da variável sexo
teste = pd.concat([teste, pd.get_dummies(teste["Embarked"])], axis=1) # Criar as variáveis Dummy a partir de Embarked
teste = teste.drop(columns=["Embarked", "S"])                         # Excluir a varuável Embarked e sua variável dummy S (Southampton)
teste = teste.replace({True: 1, False: 0})                            # Substitui os valores nas variáveis dummy por 1 e 0
teste = teste.rename(columns={'C': 'Cherbourg', "Q": 'Queenstown'})   # Trocar de nome nas colunas
teste.head(2)
```

```python
# Padronizados os dados brutos
for column in teste.columns[1:]:
    media = teste[column].mean() # Média dos valores na i-coluna
    desvio_padrao = teste[column].std() # Desvio padrão dos valores na i-coluna
    teste[column] = [(valor - media)/desvio_padrao for valor in teste[column]] # Arry com dados da i-coluna

teste.head()
```

```python
# Unificar as variáveis de entrada como seu valor de saída
teste = pd.merge(teste, survived, on='PassengerId', how='inner') 
teste.head()
```

```python
x_teste = teste[["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Cherbourg", "Queenstown"]]
x_teste.head()
```

```python
y_teste = teste["Survived"]
y_teste
```

```python
for modelo in modelos.keys():
    y_pred = modelos[modelo]["instancia"].predict(x_teste)
    scores = { "precision" : precision_score(y_teste, y_pred),
               "recall" : recall_score(y_teste, y_pred),
               "f1" : f1_score(y_teste, y_pred)
    }
    modelos[modelo]["scores"] = scores
```

```python
# Salvando todos os modelos com scores
dump(modelos, "modelos_Parte_01.joblib")
```

# Conclusão

```python
# Cria uma dicionário com os modelos e sua pontuação
resultados = {modelo : [score
                        for score in modelos[modelo]["scores"].values()]
              for modelo in modelos.keys()}

# Dataframe com modelos e sua pontuação
resultados = pd.DataFrame(resultados, index = ["precision", "recall", "f1"])
resultados
```

Analisando as métricas, podemos concluir que o melhor modelo para predição da sobrevivência do passageiro corresponde ao SVM, por ter os melhores resultados independende as métrica selecionada.

## Anexo B - Diferença entre `==` e `is` e `None` e `NaN`

1. `None`: É um objeto nativo do Python (tipo NoneType) que representa o valor "nulo" ou a ausência total de um valor. É o padrão do Python para indicar que algo não existe.
2. `NaN` (Not a Number): É um valor de ponto flutuante (tipo float) implementado pelo NumPy. Ele é usado especificamente para representar dados numéricos ausentes ou resultados de operações matemáticas indefinidas (como 0/0). 

||Origem|Uso Principal|Performace|Comparação|
|:----:|----------------------------|------------------|------|--|
|`None`|Nativo do python (`NoneType`)|Estrutura de dados|Lento |`None == None` é `true`|
|`NaN` |Biblioteca Numpy (`float`)  |Cálculo científico|Rápido|`NaN == NaN` é `False`|

3. `==` (Igualdade): Compara se o conteúdo ou os valores de dois objetos são iguais.
4. `is` (Identidade): Verifica se duas variáveis apontam para o exato mesmo objeto no mesmo endereço de memória. 

||Compara|Característica|Exemplo|
|---|---|---|---|
|`==`|O valor dos objetos|Operado de igualdade|`[1, 2] == [1, 2]` retorna `True`|
|`is`|O endereço de memória|Operador de identidade|`[1, 2] is [1, 2]` retorna `False`|

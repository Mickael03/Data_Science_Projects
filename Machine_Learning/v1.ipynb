## Variável Dummy

Uma variável dummy é utilizada para representar variáveis categóricas por meio de valores binários (0 e 1). Nesse caso, empregaremos esse tipo de variável para codificar as indicações de sexo da seguinte maneira:
|Sexo|Valor|
|----|-----|
|female|0|
|male|1|

```python
# Trocar de valores para variável categórica 'Sex'
for i in range(len(df)):
    if df.loc[i,'Sex'] == 'female':
        df.loc[i,'Sex'] = 0
    else:
        df.loc[i,'Sex'] = 1

# Mostra o dataframe
df.head()
```

Para `Embarked` transformaremos em uma variável dummy, como `Embarked` apresenta três categorias devemos criar duas variáveis dummys:
1. `Cherbourg` com valores 0 ou 1, sendo 1 quando `Embarked` tivesse sindo classificado como `C` e 0 quando não recebeu esse valor.
2. `Queenstown` com valor 0 ou 1, sendo 1 quando `Embarked` tivesse sindo classificado como `Q`e 0 quando não recebeu esse valor.
3. Quando `Cherbourg` e `Queenstown` receberem valor 0, saberemos que `Embarked` foi classificado como `S`.

Quando temos $k$ categorias, são necessários $k-1$ categorias variáveis dummy. Isso é necessários para evitamos ***Multicolinearidade*** que é quando duas ou mais variáveis independentes apresentam uma alta correlação.

****Tabela de Representação****
|Categorias|Cherbourg|Queenstown|
|:--------:|:-------:|:--------:|
|`C`|1|0|
|`Q`|0|1|
|`S`|0|0|

```python
# Lista de lista para transformar 'Embarked' em um variável Dummy
embarked = []

# Determinar o valor de cada variável Dummy
for index in range(len(df)):
    if df.loc[index,'Embarked'] == 'C':
        embarked.append([1,0])
    elif df.loc[index,'Embarked'] == 'Q':
        embarked.append([0,1])
    else:
        embarked.append([0,0])

# Dataframe como o atributo 'Embarked' transformando em um variável dummy de acordo com suas categórias
embarked = pd.DataFrame(embarked, columns=['Cherbourg','Queenstown'])

# Mostra o final dataframe
embarked.tail()
```

```python
# Concatenando os dois dataframe embarked e df
df = pd.concat([df,embarked], axis=1)

# Mostrar o Dataframe
df
```

```python
# Excluir o Atributo 'Embarked' que não é mais necessário foi trocado pelos atributos 'Cherbourg', 'Queenstown' e 'Southampton'
df.drop(columns=['Embarked'], inplace=True)

# Mostra o dataframe
df
```

```python
def ausentes(DataFrame, quantidade=False):
    # Identificação da posição dos valores ausentes
    '''
    Paramêtros:
        DataFrame: DataFrame do Pandas
        quantidade (bool): True mostra indica a quantidade de dados ausentes

    Return:
        list: Uma lista de tuplas onde temos (linha,coluna) de cada valor ausente
        int: Quantidade de dados ausentes
    '''
    ausentes = [] # Armazenar a posição da cada valor ausente
    for index in range(len(df.index)):
        for columns in range(len(df.columns)):
            if pd.isna(df.iloc[index,columns]):
                ausentes.append((index,columns))

    return len(ausentes) if quantidade else ausentes
```

```python
# Lista de valores ausentes
valores_ausentes = ausentes(df)

# Indica a quantidade de valores ausentes
print(f'Quantidade de valores ausentes: {len(valores_ausentes)}')

# Exclusão de toda as amostras com valores ausentes restantes
for posicao in valores_ausentes:
    try:  # Bloco try-except é para evitar quebrar de código já que tera index repetidos
        df.drop(posicao[0], inplace=True)
    except:
        continue

# Atualização dos index
df.index = df.index

# Mostrar o dataframe
df.head()
```

## Normalização

Sendo $X$ o valor que será transformado, $X_{Min}$ o menor valor da série dos dados e $X_{Max}$ o maior valor da série de dados, a normalização do valor $X$ será dado por:
    
<center> $X^{'}=\displaystyle\frac{X-X_{Min}}{X_{Max}-X_{Min}}$ <center>

```python
# Dataframe com valores normalizados
normalizados = pd.DataFrame()
```

```python
# Inserir valores normalizados
for i in df.columns:
    i_max = df[i].max() # Maior valor da i-coluna
    i_min = df[i].min() # Menor valor da i-coluna
    normalizados[i] = [(valor - i_min)/(i_max - i_min) for valor in df[i]] # Arry com dados da i-coluna
```

```python
# Salvado os dados normalizados e padronizados
normalizados.to_csv("Dados\\normalizados.csv", index = False)
```

## Padronização

Sendo $X$ o valor que será transfornado, $\mu$ a média e $\sigma$ o desvio padrão dos dados, a padronização do valor $X$ será dados por:

<center>$z = \displaystyle\frac{X - \mu}{\sigma}$<center>

Onde o valor $z$ é chamado de z-score.

```python
# Dataframe com valores padronizados
padronizados = pd.DataFrame()
```

```python
# Inserir valores padronizados
for i in df.columns:
    media = df[i].mean() # Média dos valores na i-coluna
    desvio_padrao = df[i].std() # Desvio padrão dos valores na i-coluna
    padronizados[i] = [(valor - media)/desvio_padrao for valor in df[i]] # Arry com dados da i-coluna
```

```python
# Salvado os dados padronizados
padronizados.to_csv("Dados\\padronizados.csv", index = False)
```

## Função de Transformação

Como dispomos de duas bases de dados:

> 1. Uma destinada à criação do modelo
> 2. Outra destinada à avaliação do modelo

A base de dados utilizada para avaliação apresenta exatamente a mesma estrutura e organização da base usada na criação do modelo. Por consequência, antes de avaliarmos o modelo diretamente pelo sklearn, devemos aplicar o mesmo tratamento de limpeza e organização dos dados. Para isso, criaremos uma função em Python que receba a base de dados bruta, execute toda a limpeza necessária e retorne um DataFrame pronto para ser utilizado tanto no treinamento quanto na avaliação do modelo.

```python
def limpeza(dados, padronizados=False, normalizados=False):
    '''
    Paramêtros:
        Dados: DataFrame pandas de dados brutos
        padronizados: Se True retorn o dataframe com dados padronizados
        normalizados: Se True retorn o dataframe com dados normalizados

    Return:
        Dataframe do pandas
    '''
   
    # Excluir os Atributos 'PassengerId', 'Survived', 'Name' e 'Ticket' do dataframe original
    dados = dados.drop(columns=['PassengerId', 'Name', 'Ticket'])

    # Excluir os dados do Atributo 'Cabin' por cerca de 77% dos dados na treinamento do modelo serem NaN
    dados = dados.drop(columns=['Cabin'])

    # Remove linhas que tenham pelo menos um NaN
    dados = dados.dropna()

    # Organizado os index
    dados.index = range(len(dados))

    # Transformação da variável sexo
    for index in range(len(dados)):
        if dados.loc[index, "Sex"] == "male":
            dados.loc[index, "Sex"] = 1
        else:
            dados.loc[index, "Sex"] = 0
  
    # Lista para variáveis dummy 'Cherbourg' e 'Queenstown'
    cherbourg, queenstown = [], []
    
    # Determinar o valor de cada variável Dummy
    for index in range(len(dados)):
        if dados.loc[index,'Embarked'] == 'C':
            cherbourg.append(1)
            queenstown.append(0)
        elif dados.loc[index,'Embarked'] == 'Q':
            cherbourg.append(0)
            queenstown.append(1)
        else: # Quanto aparece o valor `S`
            cherbourg.append(0)
            queenstown.append(0)

    # Adiciona os atributos 'Cherbourg' e 'Queenstown' 
    dados['cherbourg'] = cherbourg
    dados['queenstown'] = queenstown
    
    # Excluir o Atributo 'Embarked' que não é mais necessário
    dados = dados.drop(columns=['Embarked'])

    # Determinar o "Tipo de dataframe"
    if padronizados:
        dados = (dados - dados.mean())/dados.std() # padronizado
        return dados
    elif normalizados:
        dados = (dados - dados.min())/(dados.max() - dados.min()) # Normalizado
        return dados
    else: # Retorna os dados já limpos
        return dados
```

De outra forma, qualquer algoritmo de aprendizagem se mostrar será excelente/excepicional em determinado problema, porém mediocre em outro, de forma que na média todos os algoritmos teram o mesmo desempenhom quando consideramos todos os problemas possíveis que algoritmos podem se aplicados.

## Anexo B - Diferença entre `==` e `is` e `None` e `NaN`

1. `None`: É um objeto nativo do Python (tipo NoneType) que representa o valor "nulo" ou a ausência total de um valor. É o padrão do Python para indicar que algo não existe.
2. `NaN` (Not a Number): É um valor de ponto flutuante (tipo float) implementado pelo NumPy. Ele é usado especificamente para representar dados numéricos ausentes ou resultados de operações matemáticas indefinidas (como 0/0). 

||Origem|Uso Principal|Performace|Comparação|
|:----:|----------------------------|------------------|------|--|
|`None`|Nativo do python (`NoneType`)|Estrutura de dados|Lento |`None == None` é `true`|
|`NaN` |Biblioteca Numpy (`float`)  |Cálculo científico|Rápido|`NaN == NaN` é `False`|

3. `==` (Igualdade): Compara se o conteúdo ou os valores de dois objetos são iguais.
4. `is` (Identidade): Verifica se duas variáveis apontam para o exato mesmo objeto no mesmo endereço de memória. 

||Compara|Característica|Exemplo|
|---|---|---|---|
|`==`|O valor dos objetos|Operado de igualdade|`[1, 2] == [1, 2]` retorna `True`|
|`is`|O endereço de memória|Operador de identidade|`[1, 2] is [1, 2]` retorna `False`|
